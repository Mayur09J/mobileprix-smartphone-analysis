```python
def extended_memory_extractor(row):

  if row['extended_memory_available'] == 0:
    return np.nan
  else:
    if row['extended_memory']=='2 TB':
      return 2048
    if row['extended_memory']=='1.5 TB':
      return 1536
    if row['extended_memory']=='1 TB':
      return 1024
    if row['extended_memory']=='512 GB':
      return 512
    if row['extended_memory']=='256 GB':
      return 256
    if row['extended_memory']=='No Specified':
      return np.nan
    if row['extended_memory']=='Memory Card (Hybrid)':
     return np.nan
    if row['extended_memory']=='128 GB':
      return 128
    if row['extended_memory']=='64 GB':
      return 64
    if row['extended_memory']=='Supports MicroSD slot':
      return 1
```

```python
def core_convertor(text):

  match text:
    case _ if 'Deca Core' in text:
      return 10

    case _ if 'Nine Core' in text:
      return 9

    case _ if 'Octa Core' in text:
      return 8

    case _ if 'Hexa Core' in text:
      return  6

    case _ if 'Quad Core' in text:
      return 4

    case _ if 'Single Core' in text:
      return 1

    case _ if 'Dual Core' in text:
      return 2

    case _:
      return text

```

```python
from pandas import core
def cleaning_num_cores(core):


  match core:
    case _ if 'Octa Core Processor' in core:
      return 'Octa Core'

    case _ if 'Hexa Core Processor' in core:
      return 'Hexa Core'

    case _ if '512 MHz Processor' in core or 'Single Core Processor' in core or '1.7 GHz Processor' in core or '2 GHz Processor' in core or '2.1 GHz Processor' in core:
      return 'Single Core'

    case _ if 'Nine-Cores' in core or 'Nine Cores' in core:
      return 'Nine Core'

    case _:
      return  core




```

```python
df1['num_cores']=df1.num_cores.apply(cleaning_num_cores).str.strip()
```

```python
df1['num_cores'].value_counts()
```

```python
df1['num_cores']=df1.num_cores.apply(core_convertor)
```
