```python
import pandas as pd
from bs4 import BeautifulSoup
import numpy as np
```

```python
with open('mobile_prix_complete.html','r',encoding='utf-8')as f:
  html=f.read()
```

```python
soup=BeautifulSoup(html,'lxml')
```

```python
phones=soup.find_all('div',{'class':'sm-product has-tag has-features has-actions'})
```

```python
name=[]
price=[]
specs=[]
sim=[]
processor=[]
ram=[]
battery=[]
display=[]
camera=[]
card=[]
os=[]

for i in phones:
  try:
    name.append(i.find('h2').get_text())
  except:
    name.append(np.nan)

  try:
    price.append(i.find('span',{'class':"price"}).get_text())
  except:
    price.append(np.nan)

  try:
    specs.append(i.find('div',{'class':'score rank-2-bg'}).find('b').get_text())
  except :
    specs.append(np.nan)

  x=i.find('ul',{'class':'sm-feat specs'}).find_all('li')
  try:
    sim.append(x[0].get_text())
  except :
    sim.append(np.nan)

  try:
    processor.append(x[1].get_text())
  except :
    processor.append(np.nan)

  try:
    ram.append(x[2].get_text())
  except:
    ram.append(np.nan)

  try:
    battery.append(x[3].get_text())
  except:
    battery.append(np.nan)

  try:
    display.append(x[4].get_text())
  except:
    display.append(np.nan)

  try:
    camera.append(x[5].get_text())
  except:
    camera.append(np.nan)

  try:
    card.append(x[6].get_text())
  except:
    card.append(np.nan)

  try:
    os.append(x[7].get_text())
  except:
    os.append(np.nan)
```

```python
df=pd.DataFrame({
  'name':name,
  'price':price,
  'specs':specs,
  'sim':sim,
  'processor':processor,
  'ram':ram,
  'battery':battery,
  'display':display,
  'camera':camera,
  'card':card,
  'os':os
})
```

```python
df.to_csv('Mobile_prix_data.csv',index=False)
```
